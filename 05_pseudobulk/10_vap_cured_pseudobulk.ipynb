```python
sc_labels = pd.read_csv(common_data.SC_LABELS, index_col=0)
```

```python
BASE = common_data.DATA / '05_pseudobulk/10_vap_cured'
```

```python
labels = sc_labels.copy()
labels['group'] = labels.is_episode_cured.copy()
labels = labels.loc[labels.group.ne(-1)]
labels['group'] = labels['group'].astype(int)
labels[['bal_barcode', 'group']].dropna().to_csv(BASE / '_group_labels.csv')
labels.group.value_counts()
```
