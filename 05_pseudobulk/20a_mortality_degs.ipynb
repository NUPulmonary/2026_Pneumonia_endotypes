# Save pseudobulk labels for specific DEG tasks

- Alive vs Dead

```python
sc_labels = pd.read_csv(common_data.SC_LABELS, index_col=0)
```

```python
sc_labels.Binary_outcome.unique()
```

```python
BASE = common_data.DATA / '05_pseudobulk/20a_alive_vs_dead'
```

## Total

```python
dir = BASE / 'alive_vs_dead'
os.makedirs(dir, exist_ok=True)
death_labels = sc_labels.copy()
death_labels['group'] = death_labels.Binary_outcome.copy()
death_labels = death_labels.dropna(subset='group')
death_labels['group'] = death_labels['group'].astype(int)
death_labels[['bal_barcode', 'group']].dropna().to_csv(dir / '_group_labels.csv')
death_labels.group.value_counts()
```

### Inside viral

```python
dir = BASE / 'alive_vs_dead_viral'
os.makedirs(dir, exist_ok=True)
death_labels = sc_labels.copy()
idx = sc_labels.perturbation_groups_2.eq('Early SARS-CoV-2')
death_labels = death_labels[idx].copy()
death_labels['group'] = death_labels.Binary_outcome.copy()
death_labels = death_labels.dropna(subset='group')
death_labels['group'] = death_labels['group'].astype(int)
death_labels[['bal_barcode', 'group']].dropna().to_csv(dir / '_group_labels.csv')
death_labels.group.value_counts()
```

### Inside bacterial

```python
dir = BASE / 'alive_vs_dead_bacterial'
os.makedirs(dir, exist_ok=True)
death_labels = sc_labels.copy()
idx = sc_labels.perturbation_groups_2.isin([
    'Gram-*',
    'Gram+',
    'Pseudomonas aeruginosa',
    'Gram-*; Gram+'
])
death_labels = death_labels[idx].copy()
death_labels['group'] = death_labels.Binary_outcome.copy()
death_labels = death_labels.dropna(subset='group')
death_labels['group'] = death_labels['group'].astype(int)
death_labels[['bal_barcode', 'group']].dropna().to_csv(dir / '_group_labels.csv')
death_labels.group.value_counts()
```

## Inside pathogen negative

But select pathogen-negative samples from pneumonia episodes

```python
dir = BASE / 'alive_vs_dead_pathogen_negative'
os.makedirs(dir, exist_ok=True)
death_labels = sc_labels.copy()
death_labels = death_labels[sc_labels.is_culture_negative_pneumonia.fillna(False)].copy()
death_labels['group'] = death_labels.Binary_outcome.copy()
death_labels = death_labels.dropna(subset='group')
death_labels['group'] = death_labels['group'].astype(int)
death_labels[['bal_barcode', 'group']].dropna().to_csv(dir / '_group_labels.csv')
death_labels.group.value_counts()
```
