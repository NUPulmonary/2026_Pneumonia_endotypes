```python
import pathlib

import pandas as pd
```

```python
TO_MERGE = ['13', '20d', '31c', '40d', '50c']
```

```python
BASE = pathlib.Path('70_degs')
```

```python
for suffix in ('counts', 'stats'):
    data = []
    for prefix in TO_MERGE:
        item = pd.read_csv(BASE / f'{prefix}_deg_{suffix}.csv', index_col=0)
        data.append(item)
    data = pd.concat(data)
    if 'Perivascular_macrophages' in data.columns:
        data = data.rename(columns={'Perivascular_macrophages': 'Interstitial_macrophages'})
    else:
        data.cell_type = data.cell_type.replace({'Perivascular_macrophages': 'Interstitial_macrophages'})
    data.to_csv(BASE / f'merged_{suffix}.csv')
```
